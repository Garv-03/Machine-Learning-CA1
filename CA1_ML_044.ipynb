```python
import numpy as np

np.random.seed(42)
```

```python
A = np.random.randint(1, 11, size=(4, 3))
B = np.random.randint(1, 11, size=(3, 5))

print("Matrix A:")
print(A)
print("Shape of A:", A.shape)

print("\nMatrix B:")
print(B)
print("Shape of B:", B.shape)
```

```python
assert A.shape[1] == B.shape[0], \
    "Matrix dimensions are not compatible for multiplication."

print("Dimension validation successful.")
print("Columns of A:", A.shape[1])
print("Rows of B:", B.shape[0])
```

```python
C = np.dot(A, B)

print("Matrix C = A dot B:")
print(C)

print("\nShape of C:", C.shape)

assert C.shape == (4, 5), "Unexpected output matrix shape."
```

```python
total_sum = np.sum(C)

print("Total sum of all elements in C:", total_sum)
```

```python
row_sums = np.sum(C, axis=1)

print("Row-wise sums:")
print(row_sums)
```

```python
column_sums = np.sum(C, axis=0)

print("Column-wise sums:")
print(column_sums)
```
